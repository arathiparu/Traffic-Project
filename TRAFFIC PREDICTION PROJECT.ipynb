# LINEAR REGRESSION

```python
#DROPPING UNNECESSARY COLUMNS
```

```python
dataset.drop(["DateTime"],axis=1,inplace=True)
```

```python
dataset
```

```python
#SPLITING FEATURES INTO INDEPENDENT AND DEPENDENT VARIABLES (X,Y)
```

```python
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,1].values
```

```python
x
```

```python
y
```

```python
#SPLITTING THE DATASET INTO TRAINING AND TEST SET
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
x_test
```

```python
y_test
```

```python
#IMPORTING LINEAR REGRESSION
```

```python
from sklearn.linear_model import LinearRegression 
regressor=LinearRegression()
regressor.fit(x_train,y_train)
```

```python
#PREDICTION OF THE TEST AND TRAINING SET RESULT
```

```python
y_pred=regressor.predict(x_test)
x_pred=regressor.predict(x_train)
```

```python
#DISPLAY BOTH PREDICTIONS
```

```python
y_pred
```

```python
x_pred
```

```python
#CHECKING ACCURACY OF TRAINING AND TEST DATA
```

```python
linear_train_acc=regressor.score(x_train,y_train)
linear_test_acc=regressor.score(x_test,y_test)
print("TRAINING DATA ACCURACY:",linear_train_acc)
print("TESTING DATA ACCURACY:",linear_test_acc)
```

```python
#OUTPUT PREDICTION 
```

```python
inputs=[[4,30]]
status=regressor.predict(inputs)
print(status)
if(status>=10):
    print("THE JUNCTION HAVE TRAFFIC")
else:
    print("THE JUNCTION HAVE NO TRAFFIC")
```

# SVM

```python
#SPLITTING FEATURES INTO INDEPENDENT AND DEPENDENT VARIABLES(X,Y)
```

```python
x=dataset.iloc[:,:-1].values
y=dataset.iloc[:,1].values
```

```python
x
```

```python
#SPLITTING DATA INTO TRAINING DATA AND TESTING DATA
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
x_test
```

```python
x_pred
```

```python
svm_train_acc=classifier.score(x_train,y_train)
svm_test_acc=classifier.score(x_test,y_test)
print("TRAINING DATA ACCURACY:",svm_train_acc)
print("TESTING DATA ACCURACY:",svm_test_acc)
```

```python
inputs=[[1,9]]
status=classifier.predict(inputs)
print(status)
if(status>=10):
    print("THE JUNCTION HAVE TRAFFIC")
else:
    print("THE JUNCTION HAVE NO TRAFFIC")
```
